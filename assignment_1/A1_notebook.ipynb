#### Question-1: Number guessing game. ####
Draw a random integer from 1 to 20 and let the user guess until correct, answering “higher” or “lower” after each wrong guess.

```python
import random

def play():
    # random ile if clause kod bloklarim calisiyordu. ama user input ile calismadilar. bu yuzden komple basa dondum:
    pick = random.randint(1, 20)
    guess = 0
    hak = 1

    # senaryoyu yaziyorum
    while guess != pick:
        try:
            guess = int(input("1 ile 20 arasında bir sayı seç: "))
        except ValueError: #baska value olursa diye
            print("Lütfen geçerli bir sayı girin.")
            continue 
        except (EOFError, KeyboardInterrupt):
            print("\nOyun iptal edildi.")
            break
            
        if guess > pick:
            print("Lower")
            hak += 1
        elif guess < pick:
            print("Higher")
            hak += 1
        else: #guess == pick: ile olduramadim
            print("Congrats")
            break

# kernel cokmemesi icin oyun cancel edilme durumunu ekledim. 
play()
```

#### Question-4: Single number. 
In an array where every element appears twice except one, find that one.  


```python
#mesela 
liste = [1, 2, 3, 4, 5, 1, 3, 2, 4, 6, 1, 4, 6, 2]
#olsun

for num in liste:
    if liste.count(num) > 1:
        print(f"{num} is repeated {liste.count(num)} times")
    else:
        print("Single Number is", num)
        break
```

#### Question-5: Repeated DNA sequences. 
Return all 10-letter substrings that occur more than once in a DNA string ("AAAAACCCCCAAAAACCCCCCAAAAAGGGTTT" → ["AAAAACCCCC", "CCCCCAAAAA"]). 

```python
dna = "AAAAACCCCCAAAAACCCCCCAAAAAGGGTTT"
seen = set()
repeated = set()

#list icinde yazinca error veriyor dna = "AAAAACCCCCAAAAACCCCCCAAAAAGGGTTT" seklinde denedim
```

```python
for i in range(len(dna)-10):
    substr = dna[i:i+10]
    if substr in seen:
        repeated.add(substr)
    else:
        seen.add(substr)

print(list(repeated))

#Adim adim kontrol etmek icin 2 blokta yazdim.
```
